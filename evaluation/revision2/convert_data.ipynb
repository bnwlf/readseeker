## Data Imports

```python
import glob, re,os

from pathlib import Path
import tensorflow as tf
import tensorflow_addons as tfa

import numpy as np
from genomic_benchmarks.loc2seq import download_dataset
from genomic_benchmarks.data_check import is_downloaded, info
from genomic_benchmarks.models.tf import vectorize_layer
from genomic_benchmarks.models.tf import get_basic_cnn_model_v0 as get_model
```

```python



data = [x for x in  glob.glob("../benchmark/*/*/*_reads.fasta") if not x.startswith("../benchmark/demo_coding_vs_intergenomic_seqs")]

pattern = re.compile(r'../benchmark/([^/]+)/([^/]+)/([^/]+)_reads\.fasta')

c = {"0":"intergenomic_seqs","1":"coding_seqs"}

for elem in data:
    counter = 0
    ref,sample, _  =pattern.search(elem).groups()

    path = f"../genomic_benchmark_network_test/data/{ref}_{sample}/test/%s/%d.txt"
    os.makedirs(f"../genomic_benchmark_network_test/data/{ref}_{sample}/test/intergenomic_seqs/",exist_ok =True)
    os.makedirs(f"../genomic_benchmark_network_test/data/{ref}_{sample}/test/coding_seqs/",exist_ok =True)
    print (path%("--",counter))
    with open (elem,"r") as fasta:
        
        header = fasta.readline()
        sequence= fasta.readline()
        while sequence:
            counter += 1
            with open(path%(c[header[-2]],counter),"w") as out:
                out.write(sequence.strip())
            header = fasta.readline()
            sequence= fasta.readline()
    
    


```
